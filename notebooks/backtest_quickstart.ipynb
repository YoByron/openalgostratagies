# Backtest Quickstart
Run a minimal VectorBT/QuantStats workflow using the project helpers.

## Prerequisites
Make sure optional dependencies are installed before executing the code cells:

```bash
python -m pip install vectorbt quantstats
```

````python
import importlib
from pathlib import Path

missing = [pkg for pkg in ("vectorbt", "quantstats") if importlib.util.find_spec(pkg) is None]
if missing:
    raise RuntimeError(f"Install optional packages before running this notebook: {missing}")

import pandas as pd
from src.backtesting.backtest_runner import BacktestRunner, BacktestConfig

prices = pd.Series([100, 102, 101, 104, 107, 105], index=pd.date_range('2024-01-01', periods=6, freq='D'))

def simple_signals(price_series: pd.Series):
    entries = pd.Series(False, index=price_series.index)
    exits = pd.Series(False, index=price_series.index)
    entries.iloc[1] = True
    exits.iloc[-1] = True
    return {"prices": price_series, "entries": entries, "exits": exits}

report_dir = Path('reports/notebook_demo')
config = BacktestConfig(initial_capital=100_000, fees=0.0005, report_dir=report_dir)
runner = BacktestRunner(config)
result = runner.run(prices, simple_signals)
report = runner.build_report(result, name='notebook_quickstart', export_csv=True)

report.stats.head()
````
